```python
import sympy as sp
sp.init_printing(use_unicode=True)
```

```python
c = sp.symbols('c')
u0, u1, u2, u3 = sp.symbols('u0 u1 u2 u3')
```

```python
# Known variables
sigma11 = u2 - u1**2 / u0
sigma00 = u0
sigma12 = u3 - u2 * u1 / u0
sigma01 = u1
sigmam10 = 0

alpha1 = sigma12 / sigma11 - sigma01 / sigma00
alpha0 = sigma01 / sigma00
beta1 = sigma11 / sigma00
```

```python
# Apply the closure
u4 = (sp.Matrix([u2, u3]).T * sp.Matrix([[u0, u1], [u1, u2]]).inv() * sp.Matrix([u2, u3]))[0] + sigma11 * (2 * beta1 + (alpha1 - alpha0)**2)
```

```python
# Derivatives of the closure with respect to the moments
u4_u0 = sp.diff(u4, u0)
u4_u1 = sp.diff(u4, u1)
u4_u2 = sp.diff(u4, u2)
u4_u3 = sp.diff(u4, u3)
```

```python
# Characteristic polynomial
P4 = c**4 - u4_u0 * c**0 - u4_u1 * c**1 - u4_u2 * c**2 - u4_u3 * c**3
P4 = sp.expand(P4)
```

```python
# Replace the first moments with the variables and check eq. 
rho = u0
v = u1 / u0
theta = (u2 - rho * v**2) / rho
mu3 = (u0**2*u3 - 3*u0*u1*u2 + 2*u1**3)/u0**3

u4_tilde = rho * (v**4 + 6 * v**2 * theta + 4 * v * mu3 + 3 * theta**2 + 2 * mu3**2/theta)
```

```python
# Both closures are the same
sp.simplify(u4 - u4_tilde)
```

```python
# Calculate the derivatives of u4_tilde with respect to the moments
u4_tilde_u0 = sp.simplify(sp.diff(u4_tilde, u0))
u4_tilde_u1 = sp.simplify(sp.diff(u4_tilde, u1))
u4_tilde_u2 = sp.simplify(sp.diff(u4_tilde, u2))
u4_tilde_u3 = sp.simplify(sp.diff(u4_tilde, u3))

# Compute the characteristic polynomial using the derivatives of u4_tilde
P4_direct = c**4 - u4_tilde_u0 * c**0 - u4_tilde_u1 * c**1 - u4_tilde_u2 * c**2 - u4_tilde_u3 * c**3

sp.simplify(P4 - P4_direct)
```

```python
# Verify that eq. (90) is the true characteristic polynomial of the closure
a = mu3 / theta**sp.Rational(3, 2)
L = c * sp.sqrt(theta) + v
P4 = (L - v)**4 - (4 * mu3 / theta) * (L - v)**3 + (2 * mu3**2/theta**2 - 6*theta) * (L - v)**2 + 8*mu3 * (L - v) + 3 * theta**2

sp.simplify(P4 - P4_direct.subs(c, L))
```

```python
# Verify eq. (92) of the paper
P4_tilde = c**4 - 4*a*c**3 + (2*a**2 - 6) * c**2 + 8*a*c + 3

sp.simplify(P4 - theta**2 * P4_tilde)
```
