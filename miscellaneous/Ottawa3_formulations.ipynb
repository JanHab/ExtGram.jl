```python
import sympy as sp
sp.init_printing(use_unicode=True)
```

```python
u3 = sp.symbols('u3')
c = sp.symbols('c')
```

```python
p1 = c
p2 = c**2 - u3 * c - 1

P4 = p2**2 - (2 + u3**2) * p1**2 - 2 * (p1**2 - 1) - 2 * u3 * (p1 * p2 - 2 * p1)
sp.simplify(P4)
```

```python
beta3_tilde = (9 * u3**4 + 20 * u3**2 + 12) / (u3**2 + 2)**2
a3_tilde = 2 * u3 / (u3**2 + 2)
a2_tilde = (u3 * (3 * u3**2 + 4)) / (u3**2 + 2)
beta2 = 2 + u3**2
p3_tilde = (c - a2_tilde) * p2 - beta2 * p1
P4_tilde = (c - a3_tilde) * p3_tilde - beta3_tilde * p2

sp.simplify(P4_tilde)
```
