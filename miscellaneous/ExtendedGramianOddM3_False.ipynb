```python
import sympy as sp
sp.init_printing(use_unicode=True)
```

```python
u3 = sp.symbols('u3')
c = sp.symbols('c')
```

```python
p1 = c
p2 = c**2 - u3*c - 1
p3_hat = c**3 - u3
chi = 3/4
sigma12 = u3
sigma11 = 1

p3_tilde = p3_hat - 2 * chi * sigma12 / sigma11 * p2
sp.simplify(p3_tilde)
```

```python
P4 = p1 * p3_tilde
P4.subs({u3: 1})

roots = sp.solve(sp.Eq(P4, 0), c)
roots
```
