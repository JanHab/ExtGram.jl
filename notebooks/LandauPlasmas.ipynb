```python
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(['science', 'nature', 'grid'])

from matplotlib.lines import Line2D

import numpy as np
import pandas as pd
```

```python
# Parameter
closure = "ExtGram"
T_end = 50
Moments = [4, 8, 12, 16, 20, 24, 28, 32, 40]

x_lower = 0.0; x_upper = 12.566370614359172
base_tree_level = 8
polydeg = 1
Kn = 1.0

epsilon = 0.01
k = 0.5
rho_0 = 1.0
v_0 = 0.0
theta_0 = 1.0

markers = ['x', 'o', '^', 's', 'D', 'v']
colors = ['blue', 'orange', 'green', 'red', 'purple', 'cyan']
```

```python
gamma = -0.1533

fig, axs = plt.subplots(figsize=(6, 4))
for M in Moments:
    fig_, axs_ = plt.subplots(figsize=(6, 4))
    file = f"../out/VlasovPoisson/LandauDamping/energy_moments_solution_M{M}_closure{closure}_Kn{Kn}_T_end{T_end}_rho_0{rho_0}_v_0{v_0}_theta_0{theta_0}_epsilon{epsilon}_k{k}_p{polydeg}_level{base_tree_level}_x_lower{x_lower}_x_upper{x_upper}.csv"
    data = pd.read_csv(file)

    axs.plot(data["time"], data["E_L2"] / data["E_L2"][0], label=f"M={M}")

    axs_.plot(data["time"], data["E_L2"] / data["E_L2"][0], label=f"M={M}")
    axs_.plot(data["time"], np.exp(gamma * data["time"]), 'k--', label="theoretical decay")
    axs_.set_yscale("log")
    axs_.set_xlabel("t")
    axs_.set_ylabel("$\\bar{E}_{\\mathcal{L}_2}$")
    axs_.legend(loc='lower left')
    fig_.tight_layout()
    fig_.show()

axs.plot(data["time"], np.exp(gamma * data["time"]), 'k--', label="theoretical decay")
axs.set_yscale("log")
axs.set_xlabel("t")
axs.set_ylabel("$\\bar{E}_{\\mathcal{L}_2}$")
axs.legend(loc='lower left')
fig.tight_layout()
fig.show()
```
